```python
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, f1_score, recall_score, confusion_matrix
from sklearn.preprocessing import LabelEncoder

# Ask the user for input
n_trees = int(input("Enter the number of trees for the Random Forest: "))
impurity_function = input("Choose the impurity function ('gini' for Gini Index, 'entropy' for Information Gain): ").strip()

# Ensure correct input for impurity function
if impurity_function not in ['gini', 'entropy']:
    print("Invalid impurity function selected. Using 'gini' by default.")
    impurity_function = 'gini'

# Load the Titanic dataset from a CSV file
file_path = 'Titanic-Dataset.csv'  # Specify your file path here
titanic_data = pd.read_csv(file_path)

# Display dataset structure and first 10 rows
print("\nDataset Structure:")
print(titanic_data.info())
print("\nFirst 10 rows of the dataset:")
print(titanic_data.head(10))

# Data preprocessing: handle missing values and select relevant features
# Fill missing values in 'Age' with the median and drop rows with missing 'Embarked'
titanic_data['Age'].fillna(titanic_data['Age'].median(), inplace=True)
titanic_data.dropna(subset=['Embarked'], inplace=True)

# Drop 'Name' and 'Cabin' columns
titanic_data.drop(columns=['Name', 'Cabin'], inplace=True)

# Convert categorical features to numerical
titanic_data['Sex'] = titanic_data['Sex'].map({'male': 0, 'female': 1})
titanic_data['Embarked'] = titanic_data['Embarked'].map({'C': 0, 'Q': 1, 'S': 2})

# Create FamilySize and IsAlone features
titanic_data['FamilySize'] = titanic_data['SibSp'] + titanic_data['Parch'] + 1
titanic_data['IsAlone'] = np.where(titanic_data['FamilySize'] == 1, 1, 0)

# Categorizing the Fare attribute
fare_bins = [0, 10, 20, 30, 50, 100, 300]  # Define the bins for Fare
fare_labels = ['Very Low', 'Low', 'Medium', 'High', 'Very High', 'Luxury']  # Define labels for each bin
titanic_data['FareCategory'] = pd.cut(titanic_data['Fare'], bins=fare_bins, labels=fare_labels, right=False)

# Categorizing Age
titanic_data['AgeCategory'] = pd.cut(titanic_data['Age'], bins=[0, 12, 18, 35, 60, 100], labels=['Child', 'Teenager', 'Adult', 'Senior', 'Elder'])

# Convert categorical columns to numeric using Label Encoding
label_encoder = LabelEncoder()
titanic_data['FareCategory'] = label_encoder.fit_transform(titanic_data['FareCategory'])
titanic_data['AgeCategory'] = label_encoder.fit_transform(titanic_data['AgeCategory'])

# Select features and target variable
X = titanic_data[['Pclass', 'Sex', 'SibSp', 'Parch', 'Fare', 'Embarked', 'FamilySize', 'IsAlone', 'FareCategory', 'AgeCategory']]
y = titanic_data['Survived']

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=100)

# Create the Random Forest classifier
rf_classifier = RandomForestClassifier(n_estimators=n_trees, criterion=impurity_function, random_state=100)

# Train the Random Forest
rf_classifier.fit(X_train, y_train)

# Make predictions
y_pred = rf_classifier.predict(X_test)

# Evaluate the accuracy
accuracy = accuracy_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred, average='macro')  # macro average to handle multiple classes
recall = recall_score(y_test, y_pred, average='macro')  # same for recall
conf_matrix = confusion_matrix(y_test, y_pred)

# Display the results
print(f"\nRandom Forest accuracy: {accuracy * 100:.2f}%")
print(f"Random Forest F1-score (macro avg): {f1:.2f}")
print(f"Random Forest Recall (macro avg): {recall:.2f}")
print("Confusion Matrix:")
print(conf_matrix)
```
